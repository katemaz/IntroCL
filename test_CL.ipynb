На любом достаточно большом датасете русского языка с разметкой UD (syntagrus, например, или можно создать собственный датасет с автоматической разметкой) оцените частотность эффекта крысолова, затрагивающего инфинитив без зависимых, в относительных клаузах со словом который

```python
import pyconll
```

```python
paths = ['/Users/katyamazurina/Desktop/texts/UD/ru_syntagrus-ud-test.conllu', 
        '/Users/katyamazurina/Desktop/texts/UD/ru_syntagrus-ud-train-a.conllu',
        '/Users/katyamazurina/Desktop/texts/UD/ru_syntagrus-ud-train-b.conllu',
        '/Users/katyamazurina/Desktop/texts/UD/ru_syntagrus-ud-train-c.conllu']

#взяла размеченные датасеты из репозитрия UD Russian SynTagRus
```

```python
data = []
for path in paths:
    info = pyconll.load_from_file(path)
    data.extend(info)
```

```python
which_sentences = 0 #счетчик предложений со словом 'который'
pied_piping = 0 #счетчик предложений с эффектом крысолова 

for sentence in data:
    for token in sentence:
        if token.lemma == 'который' and token.deprel == 'acl:relcl':
            which_sentences += 1

            for token in sentence:
                if token.head == token.id and token.upos == 'VERB' and 'Inf' in token.feats.get('VerbForm', ''):
                    if not any(token.head == token.id for token in sentence):
                        pied_piping += 1  
                        break  

print(f"{which_sentences} предложений со словом 'который' ")
print(f"Эффект крысолова найден в {pied_piping} предложениях")

if which_sentences > 0:
    print(f"Частотность эффекта крысолова: {pied_piping / which_sentences * 100:.2f}%")

```
